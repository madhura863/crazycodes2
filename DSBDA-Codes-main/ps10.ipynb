# Iris Dataset Operations (Complete Notebook)
Subsetting, Merging, Sorting, Transposing, Melting, and Casting

```python
# 1. Import Library
import pandas as pd
```

## 2. Load Dataset (Fix: No Header Issue)

```python
# Load dataset without header
df = pd.read_csv("../DSBDALExam DataSets/Iris/Iris.csv", header=None)

# Assign proper column names
df.columns = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm', 'Species']

df.head()
```

```python
# Basic Info
df.info()
df.describe()
```

## g. Create data subsets for different species

```python
setosa_df = df[df['Species'] == 'Iris-setosa']
versicolor_df = df[df['Species'] == 'Iris-versicolor']

setosa_df.head(), versicolor_df.head()
```

## h. Merge two subsets

```python
merged_df = pd.concat([setosa_df, versicolor_df])
merged_df.head()
```

## i. Sort Data by Petal Length

```python
sorted_df = df.sort_values(by='PetalLengthCm', ascending=False)
sorted_df.head()
```

## j. Transposing Data

```python
transpose_df = df.T
transpose_df.head()
```

## k. Melting Data to long format

```python
melted_df = pd.melt(
    df,
    id_vars=['Species'],
    value_vars=['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm'],
    var_name='Feature',
    value_name='Value'
)

melted_df.head()
```

## l. Casting data to wide format

```python
wide_df = melted_df.pivot_table(
    index='Species',
    columns='Feature',
    values='Value',
)

wide_df
```

## Conclusion
- Subsetting done using condition
- Merging done using concat
- Sorting based on Petal Length
- Data reshaped using transpose, melt, and pivot
