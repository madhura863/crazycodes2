# Movie Dataset Analysis
Operations: Subset, Merge, Sort, Transpose, Melt, Pivot

```python
# 1. Import Library
import pandas as pd
```

```python
# 2. Load Dataset
df = pd.read_csv("../DSBDALExam DataSets/Movie/movies_metadata.csv", low_memory=False)
df.head()
```

```python
# Basic Info
df.info()
df.columns
```

## m. Subsets for different languages

```python
english_df = df[df['original_language'] == 'en']
french_df = df[df['original_language'] == 'fr']

english_df.head(), french_df.head()
```

## n. Merge subsets

```python
merged_df = pd.concat([english_df, french_df])
merged_df.head()
```

## o. Sort by customer ratings

```python
sorted_df = df.sort_values(by='vote_average', ascending=False)
sorted_df[['title', 'vote_average']].head()
```

## p. Transpose

```python
transpose_df = df.T
transpose_df.head()
```

## q. Melt to long format

```python
df['vote_average'] = pd.to_numeric(df['vote_average'], errors='coerce')
df['popularity'] = pd.to_numeric(df['popularity'], errors='coerce')
melted_df = pd.melt(
    df,
    id_vars=['original_language'],
    value_vars=['vote_average', 'popularity'],
    var_name='Metric',
    value_name='Value'
)

melted_df.head()
```

## r. Pivot to wide format

```python
df['vote_average'] = pd.to_numeric(df['vote_average'], errors='coerce')
df['popularity'] = pd.to_numeric(df['popularity'], errors='coerce')
wide_df = melted_df.pivot_table(
    index='original_language',
    columns='Metric',
    values='Value'
)

wide_df
```

## Conclusion
- Subsets created using language filter
- Merged using concat
- Sorted by ratings
- Data reshaped using transpose, melt, pivot
