# Adult Dataset (Census Bureau)
Operations: Subset, Merge, Sort, Transpose, Melt, Pivot

```python
import pandas as pd
```

## Load Dataset (Fix No Header)

```python
df = pd.read_csv("../DSBDALExam DataSets/Adult/adult.csv", header=None)

df.columns = [
    'age', 'workclass', 'fnlwgt', 'education', 'education-num',
    'marital-status', 'occupation', 'relationship', 'race', 'sex',
    'capital-gain', 'capital-loss', 'hours-per-week',
    'native-country', 'income'
]

df.head()
```

```python
df.info()
```

## s. Create subsets

```python
country_df = df[df['native-country'] == 'United-States']
male_df = df[df['sex'] == 'Male']
race_df = df[df['race'] == 'White']

country_df.head(), male_df.head(), race_df.head()
```

## t. Merge subsets

```python
merged_df = pd.concat([male_df, race_df])
merged_df.head()
```

## u. Sort Data

```python
sorted_df = df.sort_values(by='hours-per-week', ascending=False)
sorted_df.head()
```

## v. Transpose

```python
transpose_df = df.T
transpose_df.head()
```

## w. Melt

```python
melted_df = pd.melt(
    df,
    id_vars=['sex'],
    value_vars=['age', 'hours-per-week', 'education-num'],
    var_name='Feature',
    value_name='Value'
)

melted_df.head()
```

## x. Pivot

```python
wide_df = melted_df.pivot_table(
    index='sex',
    columns='Feature',
    values='Value',
    aggfunc='mean'
)

wide_df
```

## Conclusion
- Subsets created using country, sex, race
- Merged using concat
- Sorted using working hours
- Data reshaped using transpose, melt, pivot
