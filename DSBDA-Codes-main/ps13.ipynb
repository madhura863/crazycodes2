# Heart Disease Prediction (Combined Dataset)
Using Cleveland, Hungarian, Switzerland Data

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
cols = [
    'age','sex','cp','trestbps','chol','fbs','restecg',
    'thalach','exang','oldpeak','slope','ca','thal','num'
]

cleveland = pd.read_csv("../DSBDALExam DataSets/HeartDisease/Cleavland.csv", header=None)
hungarian = pd.read_csv("../DSBDALExam DataSets/HeartDisease/hung.csv", header=None)
swiss = pd.read_csv("../DSBDALExam DataSets/HeartDisease/Switzerland.csv", header=None)

cleveland.columns = cols
hungarian.columns = cols
swiss.columns = cols

df = pd.concat([cleveland, hungarian, swiss])
df.shape
```

## Data Cleaning

```python
df.replace('?', np.nan, inplace=True)
df.replace(-9, np.nan, inplace=True)

df = df.apply(pd.to_numeric, errors='coerce')
df.dropna(inplace=True)

df = df[(df >= 0).all(axis=1)]
df.shape
```

## Outlier Removal

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1

df = df[~((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))).any(axis=1)]
df.shape
```

## Data Transformation

```python
df['num'] = df['num'].apply(lambda x: 1 if x > 0 else 0)

X = df.drop('num', axis=1)
y = df['num']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

## Model Comparison

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
lr_acc = accuracy_score(y_test, lr.predict(X_test))

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
knn_acc = accuracy_score(y_test, knn.predict(X_test))

lr_acc, knn_acc
```
