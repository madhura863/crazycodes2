# Breast Cancer Prediction
Data Cleaning, Outlier Removal, Transformation, Model Comparison

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
```

## Load Dataset (Fix Header)

```python
df = pd.read_csv("../DSBDALExam DataSets/BreastCancer/BreastCancerWc.csv", header=None)

df.columns = [
    'id','clump_thickness','cell_size','cell_shape','marginal_adhesion',
    'epithelial_size','bare_nuclei','chromatin','nucleoli','mitoses','class'
]

df.head()
```

## Data Cleaning

```python
df.drop('id', axis=1, inplace=True)
df.replace('?', np.nan, inplace=True)

df = df.apply(pd.to_numeric, errors='coerce')
df.dropna(inplace=True)

df = df[(df >= 0).all(axis=1)]
df.shape
```

## Outlier Removal

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1

df = df[~((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))).any(axis=1)]
df.shape
```

## Data Transformation

```python
df['class'] = df['class'].apply(lambda x: 1 if x == 4 else 0)

X = df.drop('class', axis=1)
y = df['class']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

## Model Comparison

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(X_train, y_train)
lr_acc = accuracy_score(y_test, lr.predict(X_test))

nb = GaussianNB()
nb.fit(X_train, y_train)
nb_acc = accuracy_score(y_test, nb.predict(X_test))

lr_acc, nb_acc
```
