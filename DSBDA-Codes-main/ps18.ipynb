# Heart Disease Visualization (Set 2)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../DSBDALExam DataSets/HeartDisease/Cleavland.csv", header=None)

df.columns = [
    'age','sex','cp','trestbps','chol','fbs','restecg',
    'thalach','exang','oldpeak','slope','ca','thal','num'
]

df.head()
```

## Histogram

```python
print("Histogram Objective: To visualize distribution of Age")

plt.hist(df['age'], bins=20)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## Pie Chart

```python
# Objective: To show proportion of heart disease cases
print("Pie Chart Objective: To show proportion of Heart Disease cases")
counts = df['num'].value_counts()

labels = counts.index

plt.pie(counts, labels=labels, autopct='%1.1f%%')
plt.title("Heart Disease Distribution")
plt.show()
```

## Box Plot

```python
print("Box Plot Objective: To detect outliers in Cholesterol")

sns.boxplot(x=df['chol'])
plt.title("Box Plot of Cholesterol")
plt.xlabel("Cholesterol")
plt.show()
```

## Scatter Plot

```python
print("Scatter Plot Objective: To show relationship between Age and Cholesterol")

plt.scatter(df['age'], df['chol'])
plt.title("Scatter Plot: Age vs Cholesterol")
plt.xlabel("Age")
plt.ylabel("Cholesterol")
plt.show()
```

## Combined Plot

```python
print("Combined Plot Objective: To show relationship and distribution together")

plt.figure(figsize=(12, 5))

# Scatter Plot
plt.subplot(1, 2, 1)
plt.scatter(df['age'], df['chol'])
plt.title("Scatter Plot")
plt.xlabel("Age")
plt.ylabel("Cholesterol")

# Box Plot
plt.subplot(1, 2, 2)
sns.boxplot(y=df['chol'])
plt.title("Box Plot of Cholesterol")

plt.tight_layout()
plt.show()
```
