# Forest Fires Dataset Operations
Handling dataset without column names

```python
import pandas as pd
```

## Load Dataset (No Header)

```python
df = pd.read_csv("../DSBDALExam DataSets/forestfires/forestfires.csv")

df.head()
```

## a. Create Subsets

```python
# Objective: Classify forest fire severity based on area
print("Objective: Classify area into NotAffected, PartiallyAffected, MostlyAffected")

def classify_area(x):
    if x == 0:
        return "NotAffected"
    elif x <= 50:
        return "PartiallyAffected"
    else:
        return "MostlyAffected"

df['area_class'] = df['area'].apply(classify_area)

subset1 = df[df['area_class'] == "NotAffected"]
subset2 = df[df['area_class'] == "PartiallyAffected"]
subset3 = df[df['area_class'] == "MostlyAffected"]

print(len(subset1), len(subset2), len(subset3))
```

## b. Merge Two Subsets

```python
# Objective: Combine partially and mostly affected areas
print("Objective: Merge PartiallyAffected and MostlyAffected")

merged = pd.concat([subset2, subset3])
merged.shape
```

## c. Sort Data

```python
# Objective: Sort by temperature, wind and area
print("Objective: Sort dataset based on temp, wind and area")

sorted_df = df.sort_values(by=['temp','wind','area'])
sorted_df[['temp','wind','area']].head()
```

## d. Transpose Data

```python
# Objective: Swap rows and columns
print("Objective: Transpose dataset")

df.head().T
```

## e. Melt Data

```python
# Objective: Convert wide format to long format
print("Objective: Melt data to long format")

melted = pd.melt(df,
                 id_vars=['month','day'],
                 value_vars=['temp','RH','wind'],
                 var_name='Parameter',
                 value_name='Value')

melted.head()
```

## f. Cast to Wide Format

```python
# Objective: Convert long format back to wide format
print("Objective: Convert melted data back to wide format")

wide = melted.pivot_table(index=['month','day'],
                          columns='Parameter',
                          values='Value')

wide.head()
```
