# Hepatitis Dataset Operations
Handling dataset without column names and missing values

```python
import pandas as pd
import numpy as np
```

## Load Dataset (No Header)

```python
df = pd.read_csv("../DSBDALExam DataSets/Hepatitis/hepatitis.csv", header=None)

# Assign column names
df.columns = [
    'class','age','sex','steroid','antivirals','fatigue','malaise','anorexia',
    'liver_big','liver_firm','spleen_palpable','spiders','ascites','varices',
    'bilirubin','alk_phosphate','SGOT','albumin','protime','histology'
]

df.head()
```

## Data Cleaning

```python
# Replace '?' with NaN
df.replace('?', np.nan, inplace=True)

# Convert numeric columns
num_cols = ['age','bilirubin','alk_phosphate','SGOT','albumin','protime']

for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Drop missing values
df.dropna(inplace=True)

df.shape
```

## a. Create Subsets

```python
print("Objective: Create subsets based on sex")

male_df = df[df['sex'] == 2]
female_df = df[df['sex'] == 1]

len(male_df), len(female_df)
```

## b. Merge Subsets

```python
print("Objective: Merge subsets")

merged = pd.concat([male_df, female_df])
merged.shape
```

## c. Sort Data

```python
print("Objective: Sort by age, SGOT, protime")

sorted_df = df.sort_values(by=['age','SGOT','protime'])
sorted_df[['age','SGOT','protime']].head()
```

## d. Transpose Data

```python
print("Objective: Transpose dataset")

df.head().T
```

## e. Melt Data

```python
print("Objective: Convert wide to long format")

melted = pd.melt(df,
                 id_vars=['sex'],
                 value_vars=['age','SGOT','protime'],
                 var_name='Parameter',
                 value_name='Value')

# Convert to numeric
melted['Value'] = pd.to_numeric(melted['Value'], errors='coerce')
melted.dropna(inplace=True)

melted.head()
```

## f. Cast to Wide Format

```python
print("Objective: Convert long to wide format")

wide = melted.pivot_table(index='sex',
                          columns='Parameter',
                          values='Value')

wide.head()
```
