# E-commerce Website Review Scraper

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## Fetch Website

```python
url = "https://webscraper.io/test-sites/e-commerce/static"

response = requests.get(url)

soup = BeautifulSoup(response.text, "html.parser")

print("Website fetched successfully")
```

## Extract Product Information

```python
products = soup.find_all('div', class_='col-md-4 col-xl-4 col-lg-4')

data = []

for product in products:

    # Product Name
    name = product.find('a', class_='title')
    name = name.text.strip() if name else "N/A"

    # Price
    price = product.find('h4', class_='price')
    price = price.text.strip() if price else "N/A"

    # Description
    desc = product.find('p', class_='description')
    desc = desc.text.strip() if desc else "N/A"

    # Ratings
    stars = product.find_all('span', class_='ws-icon-star')
    rating = len(stars)

    # Reviews
    review = product.find('p', class_='review-count')
    review = review.text.strip() if review else "N/A"

    data.append({
        'Product Name': name,
        'Price': price,
        'Description': desc,
        'Rating Stars': rating,
        'Reviews': review
    })

print("Data extracted successfully")
```

## Create DataFrame

```python
df = pd.DataFrame(data)

df.head()
```

## Save CSV

```python
df.to_csv("webscraper_reviews.csv", index=False)

print("CSV file saved successfully")
```
