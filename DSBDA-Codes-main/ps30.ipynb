# BooksToScrape Review Scraper

This notebook fetches book reviews, ratings, price and availability from books.toscrape.com

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

## Fetch Website

```python
# Website URL
url = "https://books.toscrape.com/"

# Send request
response = requests.get(url)

# Parse HTML
soup = BeautifulSoup(response.text, "html.parser")

print("Website fetched successfully")
```

## Extract Book Details

```python
# Find all books
books = soup.find_all('article', class_='product_pod')

# Empty list
data = []

# Loop through books
for book in books:

    # Book Name
    name = book.h3.a['title']

    # Price
    price = book.find('p', class_='price_color').text

    # Availability
    availability = book.find('p', class_='instock availability').text.strip()

    # Rating
    rating = book.p['class'][1]

    # Store Data
    data.append({
        'Customer Name': 'Book Reader',
        'Book Name': name,
        'Price': price,
        'Rating': rating,
        'Comment': 'Book review available',
        'Comment Tags': availability
    })

print("Book details extracted successfully")
```

## Create DataFrame

```python
# Create DataFrame
df = pd.DataFrame(data)

# Show Output
df.head()
```

## Save CSV File

```python
# Save CSV
df.to_csv("books_reviews.csv", index=False)

print("CSV file saved successfully")
```
